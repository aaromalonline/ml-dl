```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder
bunch_data = load_iris()
df = pd.DataFrame(data=bunch_data.data, columns=bunch_data.feature_names)
df['CLASS'] = bunch_data.target
df.columns = ['SL', 'SW', 'PL', 'PW', 'CLASS']
print(df)
```
